```python
import sys

class CustomException(Exception):
    """
    Custom Exception Class
    - Default Python errors mein sirf error type pata hota hai
    - Ye class file name aur line number bhi deta hai
    - Debugging bahut easy ho jaati hai
    """
    
    def __init__(self, error, error_detail: sys):
        # error_detail.exc_info() se traceback object nikalo
        # 3 values aati hain — pehli 2 ki zarurat nahi, _ se ignore kiya
        _, _, exc_tb = error_detail.exc_info()
        
        # traceback se file ka naam nikalo
        file_name = exc_tb.tb_frame.f_code.co_filename
        
        # traceback se line number nikalo
        line_no = exc_tb.tb_lineno
        
        # teeno info milake ek clean message banao
        self.error_message = (
            f"Error in [{file_name}] "
            f"line [{line_no}]: "
            f"{str(error)}"
        )
    
    def __str__(self):
        # print() ya str() call karne par ye message dikhega
        return self.error_message


# agar ye file seedhi run ho (import na ho)
if __name__ == "__main__":
    try:
        a = 1/0  # intentional error — test ke liye
    except Exception as e:
        # CustomException mein error aur sys pass karo
        raise CustomException(e, sys)


# Custom Exception kyun?
# ━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━━
# ✅ File name pata hota hai — kahan error aaya
# ✅ Line number pata hota hai — exact jagah
# ✅ Error message clean format mein hota hai
# ✅ Large projects mein debugging fast hoti hai
# ✅ Production mein logs mein save hota hai
```
